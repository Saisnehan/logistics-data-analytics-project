```python
import os
import pandas as pd

data_path = "../data/data"

files = [
    "customers.csv",
    "drivers.csv",
    "trucks.csv",
    "trailers.csv",
    "facilities.csv",
    "routes.csv",
    "loads.csv",
    "trips.csv",
    "fuel_purchases.csv",
    "maintenance_records.csv",
    "delivery_events.csv",
    "safety_incidents.csv",
    "driver_monthly_metrics.csv",
    "truck_utilization_metrics.csv"
]

datasets = {}

for file in files:
    file_path = os.path.join(data_path, file)
    datasets[file.replace(".csv", "")] = pd.read_csv(file_path)

print("Datasets loaded successfully.")
```

```python
# ==========================================
# KPI ANALYSIS
# ==========================================

delivery = datasets["delivery_events"]
trips = datasets["trips"]

# KPI 1: On-Time Delivery Rate
on_time_rate = delivery["on_time_flag"].mean() * 100

# KPI 2: Average Detention Time
average_detention = delivery["detention_minutes"].mean()

# KPI 3: Total Distance
total_distance = trips["actual_distance_miles"].sum()

# KPI 4: Total Fuel Used
total_fuel = trips["fuel_gallons_used"].sum()

# KPI 5: Average Fuel Efficiency
average_mpg = trips["average_mpg"].mean()

# KPI 6: Average Idle Time
average_idle_time = trips["idle_time_hours"].mean()

# KPI 7: Average Trip Duration
average_trip_duration = trips["actual_duration_hours"].mean()


print("========== LOGISTICS KPI SUMMARY ==========")
print(f"On-Time Delivery Rate : {on_time_rate:.2f}%")
print(f"Average Detention     : {average_detention:.2f} minutes")
print(f"Total Distance        : {total_distance:,.2f} miles")
print(f"Total Fuel Used       : {total_fuel:,.2f} gallons")
print(f"Average MPG           : {average_mpg:.2f}")
print(f"Average Idle Time     : {average_idle_time:.2f} hours")
print(f"Average Trip Duration : {average_trip_duration:.2f} hours")
```

```python
# ==========================================
# ON-TIME DELIVERY BREAKDOWN
# ==========================================

delivery_counts = delivery["on_time_flag"].value_counts()

print("Delivery Event Counts:")
print(delivery_counts)

print("\nDelivery Event Percentages:")
print(delivery["on_time_flag"].value_counts(normalize=True) * 100)
```

```python
on_time_percentage = (
    delivery["on_time_flag"]
    .value_counts(normalize=True)
    .mul(100)
    .rename(index={True: "On Time", False: "Late"})
)

print(on_time_percentage)
```
